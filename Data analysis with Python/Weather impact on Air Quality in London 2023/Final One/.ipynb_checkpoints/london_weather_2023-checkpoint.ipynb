# 1. Importing dependencies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Loading the dataset

```python
df = pd.read_csv('london_weather_2023.csv')
```

# 3. Exploratory Data Analysis

# 3.1 Understanding the dataset

```python
df.describe()
```

```python
df.isna().sum()
```

```python
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing_Count': missing_count,
    'Missing_Percentage': missing_percentage
})

missing_summary = missing_summary.sort_values(by='Missing_Percentage', ascending=False)
print(missing_summary)
```

# 3.2 Cleaning the dataset

```python
print('Printing all columns before dropping:')
i = 1
for col in df.columns:
    print(i, col)
    i +=1
```

```python
# Dropping the columns with more than 50 percent data missing.
df_cols = ['preciptype', 'severerisk']
df = df.drop(columns=df_cols)
```

```python
df.isnull().sum()
```

```python
print('Printing all columns after dropping:')
i = 1
for col in df.columns:
    print(i, col)
    i +=1
```

# 3.3 Data Visualization

```python
# Convert datetime column
df['datetime'] = pd.to_datetime(df['datetime'])
df['month'] = df['datetime'].dt.month
```

```python
plt.figure(figsize=(12,5))

plt.plot(df['datetime'], df['temp'], label='Temp')
plt.plot(df['datetime'], df['tempmax'], label='Temp Max')
plt.plot(df['datetime'], df['tempmin'], label='Temp Min')

plt.xlabel('Date')
plt.ylabel('Temperature')
plt.title('Temperature Trends Over Time')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(df['datetime'], df['precip'])

plt.xlabel('Date')
plt.ylabel('Precipitation')
plt.title('Precipitation Over Time')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x='month', y='temp', data=df)

plt.xlabel('Month')
plt.ylabel('Temperature')
plt.title('Monthly Temperature Distribution')
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.scatterplot(x='temp', y='humidity', data=df, alpha=0.6)

plt.title('Temperature vs Humidity')
plt.xlabel('Temperature')
plt.ylabel('Humidity')

plt.show()
```

```python
cols = ['temp', 'humidity', 'windspeed', 'sealevelpressure', 'visibility', 'cloudcover']

for col in cols:
    plt.figure(figsize=(6,4))
    
    sns.histplot(df[col], kde=True)
    
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    
    plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(12,8))

sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```
